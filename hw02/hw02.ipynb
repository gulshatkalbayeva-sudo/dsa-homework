```python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
print(rows[1])
```

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

```python
def to_num(s):
    s = s.replace(" ", "")
    s = float(s)
    return s

def clean_city(s):
    clean_s = s.replace(" ", "").lower().capitalize()
    translation = {"Almaty":"Алматы", "Astana":"Астана", "Karaganda":"Караганда", "Shymkent":"Шымкент", "Aktobe": "Актобе", "Aturay":"Атырау"}
    latin_s = translation.get(clean_s, clean_s)
    return latin_s
```

```python
#Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной). 
 
cities = set()
for row in rows: 
    cities.add(clean_city(row['city']))
print(cities)
print(len(cities))
```

```python
#Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию. d[k] = d.get(k, 0) + 1 
cats = {}
for r in rows:
    cats[r['category']] = cats.get(r['category'], 0) + 1
print(cats)
sorted_list = sorted(cats.items(), key=lambda pair: pair[1], reverse=True)
print(sorted_list)
```

```python
#Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
total = {}
count = {}

for row in rows:
    if row["status"] == "оплачен":
        c = row["category"]
        total[c] = total.get(c, 0) + to_num(row["amount"])
        count[c] = count.get(c, 0) + 1

for c in total:
    avg = total[c] / count[c]
    print(f"{c}: выручка {total[c]}, средний чек {avg:.2f}")
```

```python
#Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`. 
months = {}

for row in rows:
    m = row["order_date"][:7]
    months[m] = months.get(m, 0) + 1

print(months)
```

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

```python
#5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
months = {}
for row in rows:
    if row["status"] == "оплачен":
        m = row["order_date"][:7]     
        months[m] = months.get(m, 0) + to_num(row["amount"])

for m in sorted(months):
    print(m, months[m])

sep = months["2026-09"]
okt = months["2026-10"]
change = (okt - sep) / sep * 100
print(f"октябрь к сентябрю: {round(change,1)}%")
```

```python
#6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
all_orders = {}
cancelled = {}

for row in rows:
    city = clean_city(row["city"])
    all_orders[city] = all_orders.get(city, 0) + 1
    if row["status"] == "отменён":
        cancelled[city] = cancelled.get(city, 0) + 1

for city in all_orders:
    share = cancelled.get(city, 0) / all_orders[city] * 100
    print(city, round(share, 1), "%")
```

```python
#7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
cust = {}
for row in rows:
    if row["status"] == "оплачен":
        c = row["customer_id"]
        cust[c] = cust.get(c, 0) + to_num(row["amount"])

top = sorted(cust.items(), key=lambda x: x[1], reverse=True)[:5]
for c, total in top:
    print(c, total)

first = top[0][0]
for row in rows:
    if row["customer_id"] == first:
        print(row)
```

Это тестовый аккаунт

```python
#8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.
city_clients = {}
for row in rows:
    city = clean_city(row["city"])
    clients = city_clients.get(city, set())
    clients.add(row["customer_id"])
    city_clients[city] = clients

for city in city_clients:
    print(city, len(city_clients[city]))
```
