```python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
```

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

```python
def to_num(s):
    s = s.replace(" ", "")
    s = float(s)
    return s

def clean_city(s):
    clean_s = s.replace(" ", "").lower().capitalize()
    translation = {"Almaty":"Алматы", "Astana":"Астана", "Karaganda":"Караганда", "Shymkent":"Шымкент", "Aktobe": "Актобе", "Aturay":"Атырау"}
    latin_s = translation.get(clean_s, clean_s)
    return latin_s
```

```python
#Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной). 
 
cities = set()
for row in rows: 
    cities.add(clean_city(row['city']))
print(cities)
print(len(cities))
```

```python
#Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
counts = {}
for row in rows:
    for c in rows:
        c = row['category']
        counts[c] = counts.get(c, 0) + 1

print(counts)
```

```python
counts.items()
sorted_list = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)
print(sorted_list)

print(f"{'Категория':} {'Выручка':} {'Ср. чек':>10}")
```

```python
#Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
total = {}
count = {}

for row in rows:
    if row["status"] == "оплачен":
        c = row["category"]
        total[c] = total.get(c, 0) + to_num(row["amount"])
        count[c] = count.get(c, 0) + 1

print(total)
print(count)
```

```python
#Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`. 
months = {}

for row in rows:
    m = row["order_date"][:7]
    months[m] = months.get(m, 0) + 1

print(months)
```
