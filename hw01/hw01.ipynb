```python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
```

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

```python
cities_clean = [c.strip().capitalize() for c in cities] #Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
print(cities_clean)

amounts_int = [int(a) for a in amounts] #Преобразуйте `amounts` из строк в числа: список `amounts_int`.
print(amounts_int)

total_orders = len(amounts_int)
total_sum = sum(amounts_int)
avg_check = round(total_sum / total_orders) #Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).

print("Заказов:", total_orders)
print("Общая сумма:", total_sum)
print("Средний чек:", avg_check)

unpaid_count = 0
unpaid_sum = 0
for amount, is_paid in zip(amounts_int, paid): # Сколько заказов не оплачено и на какую сумму?
    if not is_paid:
        unpaid_count += 1
        unpaid_sum += amount

print("Не оплачено заказов:", unpaid_count)
print("Сумма неоплаченных:", unpaid_sum)
```

```python
for city, amount, is_paid in zip(cities_clean, amounts_int, paid): #Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.
    status = "оплачен" if is_paid else "не оплачен"
    print(f"{city} | {amount} | {status}")
```

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

```python
# Выручка (только оплаченные) по Алматы и по Астане отдельно.
almaty = 0
astana = 0
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid and city == "Almaty":
        almaty += amount
    if is_paid and city == "Astana":
        astana += amount

print("Алматы:", almaty)
print("Астана:", astana)
```

```python
# Номер (индекс) и город самого крупного **оплаченного** заказа.
max_amount = 0
max_index = -1
for i in range(len(amounts_int)):
    if paid[i] and amounts_int[i] > max_amount:
        max_amount = amounts_int[i]
        max_index = i

print("Индекс:", max_index)
print("Город:", cities_clean[max_index])
print("Сумма:", max_amount)
```

```python
# Доля заказов выходного дня (`сб`, `вс`) в процентах.
weekend = 0
for d in days:
    if d == "сб" or d == "вс":
        weekend += 1

share = weekend / len(days) * 100
print(f"Доля выходных: {share:.1f}%")
```

```python
# Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 
goal = 400000
saved = 0
months = 0
while saved < goal:
    saved += 15000
    months += 1

print("Месяцев:", months, "| накоплено:", saved)
```

```python
#  А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?
saved = 0
months = 0
monthly = 15000
while saved < goal:
    saved += monthly
    months += 1
    monthly += 1000

print("Месяцев:", months, "| накоплено:", saved)
```
