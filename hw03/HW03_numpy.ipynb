````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````

````python
# Маркетинг предлагает скидку 15% на оплаченные заказы дороже 50 000 ₸. Сколько выручки магазин недополучит?
# Посчитайте дважды: со всеми заказами и без заказов клиента TEST. Почему ответы различаются в несколько раз?
sale_m = amount>50000
sum_all_m = paid[sale_m]*amount[sale_m]

sum_all_after_sale = sum_all_m*0.85
print(f"Все заказы: Выше 50000 тенге до скидки принесли бы {sum_all_m.sum()} и после {sum_all_after_sale.sum()}, теряем {sum_all_m.sum()-sum_all_after_sale.sum()} тенге")
print(f"Вся выручка магазина - {amount[paid].sum()}. Эта скидка обойдется в потерю {round((sum_all_m.sum()-sum_all_after_sale.sum())/amount[paid].sum()*100, 1)} % выручки")
test_m = cust == "TEST"
print(paid[test_m]) #all three are paid
print(amount[test_m])#all three are >50000
test_orders = amount[test_m].sum()

sum_all_wt= sum_all_m.sum() - test_orders
sum_all_wt_sale = sum_all_wt*0.85
print(f"Все заказы: Выше 50000 тенге до скидки принесли бы {sum_all_wt} и после {sum_all_wt_sale}, теряем {sum_all_wt-sum_all_wt_sale} тенге")
print(f"Вся выручка магазина - {amount[paid].sum()-test_orders}. Эта скидка обойдется в потерю {round((sum_all_wt-sum_all_wt_sale)/(amount[paid].sum()-test_orders)*100, 1)} % выручки")

print(f"Почему ответы различаются в несколько раз? \nТестовые заказы все считаются как оплаченные + каждый по 10 млн, потому и составлял большую часть заказов > 50к")
````
